## 🎯 Tujuan Praktikum:
1. Memahami penambangan pola frekuensi tinggi (*Frequent Itemsets*).
2. Membandingkan kecepatan eksekusi algoritma klasik **Apriori** dan **FP-Growth**.
3. Menghitung dan menginterpretasikan metrik: **Support**, **Confidence**, dan **Lift**.


```python
from autodatamining import AutoDataMining, load_dataset
```

## 2. Pemuatan Data Transaksi Supermarket
Daftar keranjang belanja yang memuat produk-produk yang dibeli bersamaan oleh pelanggan.


```python
transaksi = load_dataset("keranjang_belanja")
print(f"Total Transaksi: {len(transaksi)}")
for i, t in enumerate(transaksi[:5], 1):
    print(f"Transaksi {i}: {t}")
```

## 3. Eksekusi AutoDataMining (FP-Growth vs Apriori)
Sistem otomatis menguji FP-Growth dan Apriori, mencatat waktu komputasi, dan mengekstrak aturan asosiasi.


```python
results = AutoDataMining(
    task="association",
    min_support=0.25,     # Muncul di minimal 25% transaksi
    min_confidence=0.5,    # Tingkat keyakinan minimal 50%
).fit(transaksi)

# Leaderboard kecepatan algoritma
results.leaderboard
```

## 4. Frequent Itemsets & Aturan Asosiasi


```python
# Kumpulan item yang sering dibeli
print("--- Frequent Itemsets ---")
print(results.frequent_itemsets.head(10))

# Aturan asosiasi terkuat
print("\n--- Aturan Asosiasi (Diurutkan berdasarkan Lift) ---")
print(results.rules[["antecedents", "consequents", "support", "confidence", "lift"]].head(10))
```

## 5. Kesimpulan Analisis & Rekomendasi Bisnis


```python
print(results.summary())
```

## 6. Visualisasi Sebaran Aturan (Support vs Confidence)


```python
results.plot(kind="rules", show=True)
```
