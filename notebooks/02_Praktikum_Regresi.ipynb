## 🎯 Tujuan Praktikum:
1. Memahami regresi linier dan non-linier untuk memprediksi nilai kontinu.
2. Mengevaluasi metrik regresi: $R^2$, MAE (Mean Absolute Error), dan RMSE.
3. Menganalisis grafik *Residuals* dan melakukan komparasi model regresi.


```python
import pandas as pd
from autodatamining import AutoDataMining, load_dataset
```

## 2. Pemuatan Data Latihan (Harga Rumah)
Dataset berisi fitur fisik properti (luas tanah, luas bangunan, jumlah kamar, jarak pusat kota) dan harga jual dalam juta rupiah.


```python
df = load_dataset("harga_rumah")
df.head()
```

## 3. Pelatihan Model Regresi Otomatis
Bandingkan performa Linear Regression, Ridge, Lasso, Random Forest Regressor, Gradient Boosting, dll.


```python
results = AutoDataMining(
    task="regression",
    target="harga_juta",
    cv=5,
    random_state=42
).fit(df)

results.leaderboard
```

## 4. Kesimpulan Evaluasi & Draf Pembahasan Skripsi


```python
print(results.summary())
```

## 5. Visualisasi Residual
Grafik residual menunjukkan apakah galat/selisih tebakan tersebar secara acak di sekitar angka 0 (homoskedastisitas).


```python
results.plot(kind="residuals", show=True)
results.plot(kind="pipeline", show=True)
```

## 6. Format Publikasi Skripsi & LaTeX


```python
print(results.to_academic_table())
print(results.to_latex())
```
