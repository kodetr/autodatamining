```python
import pandas as pd
from autodatamining import AutoDataMining, load_dataset
```

## 2. Pemuatan Data Segmentasi Nasabah Bank
Dataset ini berisi data demografi dan keuangan nasabah (usia, pendapatan, skor pengeluaran, saldo rekening).


```python
df = load_dataset("segmentasi_nasabah")
df.head()
```

## 3. Pelatihan & Evaluasi Klaster
AutoDataMining akan menguji berbagai algoritma klaster dan mengurutkannya berdasarkan skor Silhouette.


```python
results = AutoDataMining(
    task="clustering",
    random_state=42
).fit(df)

results.leaderboard
```

## 4. Kesimpulan Karakteristik Klaster


```python
print(results.summary())
```

## 5. Visualisasi Klaster 2D (PCA Projection)


```python
results.plot(kind="clusters", show=True)
```
