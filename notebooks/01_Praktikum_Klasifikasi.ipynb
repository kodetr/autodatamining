```python
# Langkah 1: Instalasi pustaka (jika dijalankan di Google Colab)
# !pip install -q "autodatamining[visualization]"

import pandas as pd
from autodatamining import AutoDataMining, load_dataset, explain_concept
```

## 2. Pemuatan Data Latihan (Dataset Kelulusan Mahasiswa)
Dataset ini berisi data mahasiswa dengan fitur IPK, SKS, Kehadiran, dan status kelulusan tepat waktu (`Ya`/`Tidak`).


```python
# Muat dataset bawaan
df = load_dataset("kelulusan_mahasiswa")
print(f"Jumlah Baris: {len(df)}, Jumlah Kolom: {len(df.columns)}")
df.head()
```

## 3. Profiling & Pengecekan Data
Sebelum melatih model, periksa apakah terdapat data kosong (*missing values*) atau ketidakseimbangan kelas.


```python
profile = AutoDataMining.profile(df, target="lulus_tepat_waktu")
print("Potensi Masalah:", profile["potential_issues"])
print("Rincian Kolom:", profile["column_details"])
```

## 4. Pelatihan & Perbandingan Model Otomatis
AutoDataMining otomatis melakukan:
1. Prapemrosesan (Imputasi & Encoding).
2. Pemisahan data 80% latih dan 20% uji dengan stratifikasi.
3. 5-Fold Cross-Validation untuk menguji seluruh model kandidat.


```python
results = AutoDataMining(
    task="classification",
    target="lulus_tepat_waktu",
    cv=5,
    random_state=42
).fit(df)

# Tampilkan Leaderboard Peringkat Algoritma
results.leaderboard
```

## 5. Kesimpulan Naratif & Interpretasi Ilmiah
Gunakan `results.summary()` untuk membaca penjelasan performa model, deteksi overfitting, dan draf kalimat pembahasan skripsi.


```python
print(results.summary())
```

## 6. Visualisasi Pipeline & Confusion Matrix
Lihat diagram alur langkah data mining dan matriks kebingungan (*Confusion Matrix*).


```python
# Diagram alur pipeline penelitian
results.plot(kind="pipeline", show=True)

# Confusion Matrix
results.plot(kind="confusion_matrix", show=True)
```

## 7. Penyetelan Parameter Otomatis (Hyperparameter Tuning)
Mari kita optimalkan model pemenang untuk melihat peningkatan akurasi sebelum vs sesudah tuning.


```python
tuned = results.tune()
print(tuned.summary())
tuned.comparison
```

## 8. Ekspor Format Tabel Skripsi & LaTeX
Salin tabel berikut langsung ke Bab IV Skripsi atau Overleaf jurnal.


```python
print("=== Tabel Format Skripsi (Markdown) ===")
print(results.to_academic_table())

print("\n=== Kode Tabel LaTeX (Overleaf/IEEE) ===")
print(results.to_latex())
```
