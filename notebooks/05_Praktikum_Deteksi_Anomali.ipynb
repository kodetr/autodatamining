## 🎯 Tujuan Praktikum:
1. Memahami identifikasi observasi pencilan (*outliers*) yang menyimpang dari pola umum.
2. Membandingkan algoritma **Isolation Forest**, **Local Outlier Factor (LOF)**, dan **One-Class SVM**.
3. Menganalisis studi kasus deteksi transaksi keuangan mencurigakan (*fraud detection*).


```python
import pandas as pd
from autodatamining import AutoDataMining, load_dataset
```

## 2. Pemuatan Data Transaksi Keuangan
Data mencakup nominal transfer, jarak transaksi dari domisili, jam transaksi, dan frekuensi dalam 24 jam.


```python
df = load_dataset("deteksi_penipuan")
df.head()
```

## 3. Eksekusi Deteksi Anomali


```python
results = AutoDataMining(
    task="anomaly",
    models=["isolation_forest", "lof", "one_class_svm"],
    random_state=42
).fit(df)

results.leaderboard
```

## 4. Kesimpulan Anomali & Temuan Pencilan


```python
print(results.summary())
```

## 5. Visualisasi Titik Anomali (Label -1 = Anomali)


```python
results.plot(kind="anomalies", show=True)
```
